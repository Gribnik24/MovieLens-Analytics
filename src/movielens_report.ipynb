# __*Movielens Report*__

__*Current report contains information about movies that was gained with the methods from movielens_analytics.py*__

```python
from movielens_analytics import *
```

```python
movies = Movies('../data/movies.csv')
tags = Tags('../data/tags.csv')
ratings = Ratings('../data/ratings.csv')
```

```python
def dist_printer(dist_info, n=10, money_flag=False, time_flag=False):
    """
    Function for presenting the distribution in user-friendly way
    """
    
    for num, pair in dist_info:
        if num <= n:
            if money_flag:
                print(f'{pair[0]}: {pair[1]} $')
            elif time_flag:
                print(f'{pair[0]}: {pair[1]} mins')
            else:
                print(f'{pair[0]}: {pair[1]}')
        else:
            break
```

## __*Analysis of movies.csv*__


### *__Cinematic chronology__*

*__Explore how the popularity of film production has changed over the years__*

```python
dist_info = enumerate(movies.dist_by_release().items())
dist_printer(dist_info)
```

### *__Kings of genres__*

*__Explore the distribution of films by genre__*


```python
dist_info = enumerate(movies.dist_by_genres().items())
dist_printer(dist_info)
```

### *__Genre chameleons__*

*__Explore films that manage to combine many genres__*


```python
dist_info = enumerate(movies.most_genres(10).items())
dist_printer(dist_info)
```

## __*Analysis of tags.csv*__


### *__Eloquent spectators__*

*__Explore tags where users don't skimp on words.__*


```python
dist_info = enumerate(tags.most_words(10).items())
dist_printer(dist_info)
```

### *__Marathon runners by symbols__*

*__Explore the tags that hold the record for the most characters__*


```python
dist_info = tags.longest(10)
print(*dist_info, sep='\n')
```

### *__Golden Fund__*

```python
dist_info = tags.most_words_and_longest(10)
print(*dist_info, sep='\n')
```

### *__Voice of the Audience__*

*__Exploring the most popular tags used by users to describe films__*

```python
dist_info = enumerate(tags.most_popular(10).items())
dist_printer(dist_info)
```

### *__Thematic Investigation__*
*__Searching for tags by keyword: "funny"__*

```python
dist_info = tags.tags_with('funny')
print(*dist_info, sep='\n')
```
